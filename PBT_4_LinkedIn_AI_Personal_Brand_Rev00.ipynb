```python
# Cell 11 — Validate Interactive Web Links
#
# This validates link structure locally.
# It does not pretend to prove that an external service is currently online.

def valid_https_url(value):
    parsed = urlparse(value)
    return parsed.scheme == "https" and bool(parsed.netloc)

link_validation = {
    label: valid_https_url(url)
    for label, url in professional_links.items()
}

for label, passed in link_validation.items():
    print(f"{label:<38}: {'PASSED' if passed else 'FAILED'}")

external_links_validation_passed = all(link_validation.values())

print(
    "External links validation:",
    "PASSED" if external_links_validation_passed else "FAILED"
)
```
